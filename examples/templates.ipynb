# MakieTemplates — template gallery

One dummy plot per template. Requires IJulia (not a package dependency — `Pkg.add("IJulia")` in your global environment).

All templates return `(fig, ax, plot)` and take `figure=(width_fraction=..., textwidth_pt=...)` for exact `\textwidth` sizing (see README).

```julia
using Pkg; Pkg.activate(joinpath(@__DIR__, ".."))
using juliaPlotting, CairoMakie, Random

Random.seed!(1)
x = collect(range(0, 4π, 40))
y = sin.(x)
yerr = 0.1 .+ 0.1 .* rand(length(x));
```

## Scatter

```julia
fig, ax, plt = plot_xy(x, y; xlabel=L"x", ylabel=L"\sin(x)")
fig
```

## Lines

```julia
fig, ax, plt = plot_xy(x, y; markers=false, lines=true, xlabel=L"x", ylabel=L"\sin(x)")
fig
```

## Scatter + lines

```julia
fig, ax, plt = plot_xy(x, y; lines=true, xlabel=L"x", ylabel=L"\sin(x)")
fig
```

## Lines + error band

```julia
fig, ax, plt = plot_xy(x, y; yerr, markers=false, lines=true, errorstyle=:band, xlabel=L"x", ylabel=L"\sin(x)")
fig
```

## Scatter + per-point errorbars

```julia
fig, ax, plt = plot_xy(x, y; yerr, errorstyle=:bars, xlabel=L"x", ylabel=L"\sin(x)")
fig
```

## Scatter + lines + band + errorbars

```julia
fig, ax, plt = plot_xy(x, y; yerr, lines=true, errorstyle=:both, xlabel=L"x", ylabel=L"\sin(x)")
fig
```

## `\textwidth` sizing

Half-textwidth figure — include in LaTeX with `\includegraphics[width=0.5\textwidth]{scatter_halfwidth.pdf}` and the fonts come out at exactly the theme's pt sizes.

```julia
fig, ax, plt = plot_xy(x, y; xlabel=L"x", ylabel=L"\sin(x)", figure=(width_fraction=0.5, textwidth_pt=455))
fig
```

## Heatmap

```julia
xs = collect(range(-2, 2, 50))
ys = collect(range(-2, 2, 50))
z = [sin(2xi) * cos(2yi) for xi in xs, yi in ys]

fig, ax, plt = plot_heatmap(xs, ys, z; xlabel=L"x", ylabel=L"y", colorbarlabel=L"z")
fig
```

## Filled contour

```julia
fig, ax, plt = plot_contour(xs, ys, z; filled=true, xlabel=L"x", ylabel=L"y", colorbarlabel=L"z")
fig
```

## Line contour with labeled levels

```julia
fig, ax, plt = plot_contour(xs, ys, z; labels=true, xlabel=L"x", ylabel=L"y")
fig
```

## Histogram

```julia
fig, ax, plt = plot_histogram(randn(1000); normalization=:pdf, xlabel=L"v", ylabel=L"p(v)")
fig
```

## Vector field (quiver)

```julia
xq = collect(range(-2, 2, 15))
yq = collect(range(-2, 2, 15))
u = [-yj for _ in xq, yj in yq]
v = [xi for xi in xq, _ in yq]

fig, ax, plt = plot_quiver(xq, yq, u, v; xlabel=L"x", ylabel=L"y", colorbarlabel=L"|\vec{u}|")
fig
```
